# Muestras aleatorias (Monte Carlo)

## 50 muestras

```python
np.random.seed(20)

# Muestras aleatorias de U(0,1)
muestras = np.random.uniform(0, 1, 50)

# 3. Estadísticos descriptivos 
media = np.mean(muestras)
mediana = np.median(muestras)
desv_std = np.std(muestras, ddof=1)
minimo = np.min(muestras)
maximo = np.max(muestras)

print(f"Media: {media:.4f}")
print(f"Mediana: {mediana:.4f}")
print(f"Desviación Estándar: {desv_std:.4f}")
print(f"Mínimo: {minimo:.4f}")
print(f"Máximo: {maximo:.4f}")
```

```python
plt.figure(figsize=(8, 5))
plt.hist(muestras, bins=8, edgecolor='black', color='skyblue', alpha=0.7)
plt.title('Histograma de 50 muestras aleatorias U(0,1)')
plt.xlabel('Valores')
plt.ylabel('Frecuencia')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

La simulación de Monte Carlo con 50 muestras de U(0,1) nos dicen que la media muestral y la mediana se aproximan al valor teórico. mostrando una buena tendencia central. También, los valores mínimo y máximo cubren el dominio sin agruparse en los extremos, y el histograma muestra las fluctuaciones del comportamiento estocástico de las muestras.

# Sobol

```python
from scipy.stats import qmc
```

```python
# Función 
f = lambda x: x**2
valor_exacto = 1 / 3
a, b = 0, 1

m_valores = [100, 500, 1000]

print("Comparativa Monte Carlo (MC) vs Quasi-Monte Carlo (QMC - Sobol)")
print(f"{'m':<6} | {'MC: f(c)(b-a)':<14} | {'Error MC':<12} | {'QMC: f(c)(b-a)':<14}")
print("-" * 65)

np.random.seed(20)
for m in m_valores:
    # Monte Carlo 
    muestras_mc = np.random.uniform(a, b, m)
    fc_mc = np.mean(f(muestras_mc))
    aprox_mc = fc_mc * (b - a)
    error_mc = np.abs(aprox_mc - valor_exacto)
    
    # Quasi-Monte Carlo (Sobol)
    sampler = qmc.Sobol(d=1, scramble=True, seed=20)
    muestras_qmc = sampler.random(n=m).flatten()
    fc_qmc = np.mean(f(muestras_qmc))
    aprox_qmc = fc_qmc * (b - a)
    
    print(f"{m:<6} | {aprox_mc:<14.6f} | {error_mc:<12.6f} | {aprox_qmc:<14.6f}")
```

```python
mu, sigma = 50, 10
umbral = 60
valor_exacto = 1 - norm.cdf(umbral, loc=mu, scale=sigma)

m_valores = [10, 100, 1000, 10000]

print("Comparativa P(Utilidad > 60) entre MC y QMC (Sobol)")
print(f"{'m':<6} | {'MC: P(X > 60)':<14} | {'Error MC':<12} | {'QMC: P(X > 60)':<14}")
print("-" * 65)

np.random.seed(20)
for m in m_valores:
    # Monte Carlo Tradicional
    muestras_mc = np.random.normal(loc=mu, scale=sigma, size=m)
    p_mc = np.mean(muestras_mc > umbral)
    error_mc = np.abs(p_mc - valor_exacto)
    
    # Quasi-Monte Carlo (Sobol + Inversa Normal PPF)
    sampler = qmc.Sobol(d=1, scramble=True, seed=20)
    u_qmc = sampler.random(n=m).flatten()
    muestras_qmc = norm.ppf(u_qmc, loc=mu, scale=sigma)
    p_qmc = np.mean(muestras_qmc > umbral)
    
    print(f"{m:<6} | {p_mc:<14.6f} | {error_mc:<12.6f} | {p_qmc:<14.6f}")
```

# Call europea y Call asiática

## Datos

```python
So = 100
r = 0.05
sigma = 0.20
T = 1
n = 252
K = 100
M = 1024
```

## Sobol

```python
dt = T/n

# Sobol
sampler = stats.qmc.Sobol(d=n, scramble=True, seed=20)
u_muestras = sampler.random(M)

z = stats.norm.ppf(u_muestras)
```

Usamos una dimensionalidad (número de valores aleatorios para cada simulación) de 252 para Sobol, porque son los días de la caminata por simulación a considerar.

Creamos 1024 simulaciones porque el generador de Sobol funciona con potencias de 2.

## Retornos logaritmicos

Se calculan los incrementos logarítmicos diarios, la volatilidad y el tamaño del paso de tiempo. Luego, con la suma acumulativa (cumsum), se integran estos cambios diarios a lo largo de los 252 pasos para cada trayectoria, y al final se aplica la función exponencial multiplicada por el precio inicial.

```python
log_returnos = (r - 0.5 * sigma**2)* dt + sigma * np.sqrt(dt) * z
log_paths = np.cumsum(log_returnos, axis=1)
paths = So * np.exp(log_paths)
```

## Europea

Se procesa el precio del activo al vencimiento para calcular el payoff 
$$max(S_T - K, 0)$$
, cuyo valor promedio sobre las 1024 simulaciones se descuenta a valor presente con la tasa libre de riesgo. Se calcula la solución analítica con las fórmulas de Black-Scholes usando la función de distribución normal acumulada (cdf), lo que sirve para verificar la precisión del método Quasi-Monte Carlo.

```python
payoff_eur = np.maximum(paths[:, -1]-K, 0)
precio_eur_qmc = np.exp(-r*T)*np.mean(payoff_eur)
d1 = (np.log(So/K)+(r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
d2 = d1 - sigma * np.sqrt(T)
precio_eur = So * stats.norm.cdf(d1) - K * np.exp(-r * T) * stats.norm.cdf(d2)
```

## Asiática

Se obtiene el precio promedio diario de cada trayectoria a lo largo de los 252 días (np.mean por fila) para determinar el payoff $$max(prom{S} - K, 0)$$, el cual se promedia entre todas las simulaciones y se descuenta a valor presente a la tasa r.

```python
avg_precios = np.mean(paths, axis=1)
payoff_asia = np.maximum(avg_precios - K, 0)
precio_asia_mc = np.exp(-r * T) * np.mean(payoff_asia)
precio_asia_qmc = np.exp(-r*T)*np.mean(payoff_asia)
```

## Resultados Call europea y Call asiática

```python
print(f'Call europea de : {precio_eur:6f}')
print(f'Call asiatica MC: {precio_asia_mc:6f}')
print(f'Call asiatica QMC: {precio_asia_qmc:6f}')
```
